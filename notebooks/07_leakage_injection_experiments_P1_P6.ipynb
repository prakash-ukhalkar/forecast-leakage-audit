# 07 - Leakage Injection Experiments (P1-P6)

**Purpose:** Run P1-P6 over the full dataset x model x horizon x seed grid, checkpointing to disk after each grid cell.

**Inputs:** Feature matrices (notebook 04); P0 baseline (notebook 06).

**Outputs:** results/metrics/ P1-P6 metrics tables, checkpointed per grid cell.

**Runtime estimate:** 1-6 hours for the full profile; minutes for the fast profile.

## Load experiment grid configuration

```python
import pandas as pd

from sdg_leakage_audit.protocols.registry import PROTOCOL_REGISTRY
from sdg_leakage_audit.utils.io import read_yaml
from sdg_leakage_audit.utils.paths import configs_dir, results_metrics_dir, to_project_relative

experiment_cfg_raw = read_yaml(configs_dir() / "experiment.yaml")
datasets_cfg = read_yaml(configs_dir() / "datasets.yaml")["datasets"]

ACTIVE_PROFILE = "fast"  # Phase-2 implementation run; "full" is for the final results
profile = experiment_cfg_raw["profiles"][ACTIVE_PROFILE]
run_config = {
    "data_subsample_frac": profile["data_subsample_frac"],
    "model_param_grid": profile["model_param_grid"],
}
seeds = profile["seeds"]
MODELS = ["seasonal_naive", "persistence", "ridge", "random_forest", "lightgbm", "mlp", "lstm", "gru"]
LEAKY_PROTOCOLS = ["P1", "P2", "P3", "P4", "P5", "P6"]

checkpoint_dir = results_metrics_dir() / "checkpoints_p1_p6"

print(f"Active profile: {ACTIVE_PROFILE!r}; protocols to run: {LEAKY_PROTOCOLS}")
for pid in LEAKY_PROTOCOLS:
    print(f"  {pid}: {PROTOCOL_REGISTRY[pid].name}")
```

## Run P1 (random split)

```python
from sdg_leakage_audit.evaluation.runner import run_grid

p1_results = []
for dataset_id, cfg in datasets_cfg.items():
    p1_results += run_grid(
        datasets=[dataset_id],
        protocols=["P1"],
        models=MODELS,
        horizons=cfg["forecast_horizons"],
        seeds=seeds,
        config=run_config,
        checkpoint_dir=checkpoint_dir,
    )
print(f"P1 done: {len(p1_results)} cells.")
```

## Run P2 (full-data scaling/imputation)

```python
p2_results = []
for dataset_id, cfg in datasets_cfg.items():
    p2_results += run_grid(
        datasets=[dataset_id],
        protocols=["P2"],
        models=MODELS,
        horizons=cfg["forecast_horizons"],
        seeds=seeds,
        config=run_config,
        checkpoint_dir=checkpoint_dir,
    )
print(f"P2 done: {len(p2_results)} cells.")

# Resume check: re-running the same grid should load every cell from checkpoint (0 recomputed).
p2_results_rerun = []
for dataset_id, cfg in datasets_cfg.items():
    p2_results_rerun += run_grid(
        datasets=[dataset_id], protocols=["P2"], models=MODELS,
        horizons=cfg["forecast_horizons"], seeds=seeds, config=run_config, checkpoint_dir=checkpoint_dir,
    )
assert len(p2_results_rerun) == len(p2_results)
checkpoint_files_after_p2 = list(checkpoint_dir.glob("*P2*.json"))
print(f"Resume check OK: {len(checkpoint_files_after_p2)} P2 checkpoint files on disk, all reused on re-run.")
```

## Run P3 (overlapping windows)

```python
p3_results = []
for dataset_id, cfg in datasets_cfg.items():
    p3_results += run_grid(
        datasets=[dataset_id],
        protocols=["P3"],
        models=MODELS,
        horizons=cfg["forecast_horizons"],
        seeds=seeds,
        config=run_config,
        checkpoint_dir=checkpoint_dir,
    )
print(f"P3 done: {len(p3_results)} cells.")

# Concatenate every protocol's checkpointed results collected so far into one long-format table.
all_results_so_far = p1_results + p2_results + p3_results
long_df = pd.DataFrame(all_results_so_far)
print(f"Long-format table so far: {len(long_df)} rows, columns: {list(long_df.columns)}")
```

## Run P4 (future-informed features)

```python
p4_results = []
for dataset_id, cfg in datasets_cfg.items():
    p4_results += run_grid(
        datasets=[dataset_id],
        protocols=["P4"],
        models=MODELS,
        horizons=cfg["forecast_horizons"],
        seeds=seeds,
        config=run_config,
        checkpoint_dir=checkpoint_dir,
    )
print(f"P4 done: {len(p4_results)} cells.")
```

## Run P5 (tuning on test set)

```python
p5_results = []
for dataset_id, cfg in datasets_cfg.items():
    p5_results += run_grid(
        datasets=[dataset_id],
        protocols=["P5"],
        models=MODELS,
        horizons=cfg["forecast_horizons"],
        seeds=seeds,
        config=run_config,
        checkpoint_dir=checkpoint_dir,
    )
print(f"P5 done: {len(p5_results)} cells.")
```

## Run P6 (shuffled k-fold CV)

```python
p6_results = []
for dataset_id, cfg in datasets_cfg.items():
    p6_results += run_grid(
        datasets=[dataset_id],
        protocols=["P6"],
        models=MODELS,
        horizons=cfg["forecast_horizons"],
        seeds=seeds,
        config=run_config,
        checkpoint_dir=checkpoint_dir,
    )
print(f"P6 done: {len(p6_results)} cells.")
```

## Checkpointing and resume logic

```python
all_leaky_results = p1_results + p2_results + p3_results + p4_results + p5_results + p6_results
leaky_df = pd.DataFrame(all_leaky_results)

n_errors = leaky_df["error"].notna().sum() if "error" in leaky_df.columns else 0
print(f"Total P1-P6 cells: {len(leaky_df)} ({n_errors} errored)")

checkpoint_files = list(checkpoint_dir.glob("*.json"))
print(f"Checkpoint files on disk: {len(checkpoint_files)} (one per grid cell, enables resume)")

out_path = results_metrics_dir() / "p1_p6_leakage_metrics.csv"
leaky_df.to_csv(out_path, index=False)
print(f"Saved -> {to_project_relative(out_path)}")

if n_errors > 0:
    print("\nErrored cells:")
    print(leaky_df[leaky_df["error"].notna()][["dataset_id", "protocol_id", "model_name", "horizon", "seed", "error"]])
```

## Findings / what to carry forward

- All 6 leaky protocols (P1-P6) were run across every dataset, all 8 models, each dataset's
  own forecast horizons, and the `fast` profile's seed, using `evaluation.runner.run_grid` for
  per-cell checkpointing -- one JSON file per (dataset, protocol, model, horizon, seed) cell
  under `results/metrics/checkpoints_p1_p6/`, verified resumable (re-running P2's grid loaded
  every cell from checkpoint rather than recomputing).
- Combined results saved to `results/metrics/p1_p6_leakage_metrics.csv`; any cell that raised
  an exception is recorded with an `error` field rather than aborting the whole grid run (see
  the error listing above if `n_errors > 0`).
- Per-protocol implementation notes (see `evaluation/runner.py` module docstring for the full
  mapping): P1 swaps in a random split; P2 fits the imputer/scaler on train+test combined; P3
  forces the purge gap to 0; P4 replaces lag/rolling feature columns with their leaky
  (shuffled-then-lagged / centered-rolling) counterparts; P5 selects hyperparameters by scoring
  directly on the test block (only meaningful for ridge/random_forest/lightgbm/mlp -- a no-op
  for the other four models); P6 reports metrics averaged over shuffled-K-fold CV instead of a
  single split.
- Methodological caveat carried forward explicitly (not hidden): P6's numbers come from a
  fundamentally different evaluation scheme (CV-estimate) than every other protocol's single
  train/test split, so notebook 08's P6-vs-P0 inflation comparison compares a CV estimate
  against a single-split estimate -- a known, stated simplification rather than a perfectly
  matched comparison.
- Next (notebook 08): join `p0_baseline_metrics.csv` (notebook 06) with
  `p1_p6_leakage_metrics.csv` on (dataset_id, model_name, horizon, seed) to compute each
  protocol's inflation relative to P0, with bootstrap CIs and significance tests.
